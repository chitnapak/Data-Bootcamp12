```python
class ATM:
  def __init__(self, username, bank, amount):
    self.username = username
    self.bank = bank
    self.amount = amount

  def __str__(self):
    return f"{self.username} uses {self.bank} and has {self.amount} THB in an account."

  def check_balance(self):
    print(f"Balance: {self.amount} THB")

  def deposit(self, saving):
    self.amount += saving
    print(f"Deposit {saving} THB")
    print(f"Balance: {self.amount} THB")

  def withdraw(self, draw):
    self.amount -= draw
    print(f"Withdraw {saving} THB")
    print(f"Balance: {self.amount} THB")
```

```python
class ATM:
  def __init__(self, account_name, pin, balance=0):
    self.account_name = account_name
    self.pin = pin
    self.balance = balance
    self.history = []
    self.log_history(f"Account created with initial balance of ${balance}")

  def log_history(self, message):
    self.history.append(message)

  def check_balance(self, input_pin):
    if input_pin == self.pin:
      print(f"\n--- [ {self.account_name}'s Balance : ${self.balance} ] ---")
    else:
      print("Wrong pin")

  def deposit(self, amount):
    if amount > 0:
      self.balance += amount
      self.log_history(f"Deposited ${amount}")
      print(f"\n--- [ ${amount} has been deposited to {self.account_name}'s account ] ---")

  def withdraw(self, input_pin, amount):
    if input_pin != self.pin:
      print("Wrong pin")
      return

    elif amount > self.balance:
      print("Insufficient balance")

    else:
      self.balance -= amount
      self.log_history(f"Withdrew ${amount}")
      print(f"\n--- [ ${amount} has been withdrawn from {self.account_name}'s account ] ---")

  def transfer(self, input_pin, target_account, amount):
    if input_pin != self.pin:
      print("Wrong pin")
      return

    if amount > self.balance:
      print("Insufficient balance")
    else:
      self.balance -= amount
      target_account.receive_funds(amount, self.account_name)
      self.log_history(f"Transferred ${amount} to {target_account.account_name}")
      print(f"\n--- [ ${amount} has been transferred to {target_account.account_name}'s account ] ---")

  def receive_funds(self, amount, sender_name):
    self.balance += amount
    self.log_history(f"Received from {sender_name}: +${amount}")

  def get_history(self):
    print("\n--- [ Transaction History ] ---")
    for entry in self.history:
      print(entry)

  def show_history(self):
    print(f"\n---  Transaction History for: {self.account_name}  ---")
    for i, record in enumerate(self.history, 1):
      print(f"{i}. {record}")
```

```python
import random

class ATM:
    def __init__(self, account_name, pin, balance=0):
        self.account_name = account_name
        self.__pin = pin
        self.__balance = balance
        self.__current_otp = None # สำหรับเก็บ OTP ที่สุ่มได้ชั่วคราว
        self.history = []
        self.log_history(f"เปิดบัญชีใหม่ ยอดเงิน: ${balance}")

    def log_history(self, message):
        self.history.append(message)

    # --- ฟังก์ชันจัดการ OTP ---
    def __request_otp(self):
        """สุ่ม OTP 6 หลัก และจำลองการส่ง SMS"""
        self.__current_otp = str(random.randint(100000, 999999))
        print(f"\n[SMS] เรียนคุณ {self.account_name}, รหัส OTP ของคุณคือ: {self.__current_otp}")
        return self.__current_otp

    def __verify_otp(self):
        """ให้ผู้ใช้กรอก OTP และตรวจสอบความถูกต้อง"""
        user_input = input("กรุณากรอกรหัส OTP ที่ได้รับ: ")
        if user_input == self.__current_otp:
            self.__current_otp = None # ล้างค่า OTP หลังใช้งานสำเร็จ
            return True
        else:
            print("❌ OTP ไม่ถูกต้อง รายการถูกยกเลิก")
            return False

    # --- ฟังก์ชันหลักที่ต้องใช้ OTP ---
    def deposit(self, amount):
        if amount <= 0: return

        print(f"\n--- กำลังทำรายการฝากเงิน: ${amount} ---")
        self.__request_otp() # ส่ง OTP
        if self.__verify_otp(): # ตรวจสอบ OTP
            self.__balance += amount
            self.log_history(f"ฝากเงิน: +${amount}")
            print(f"✅ ฝากเงินสำเร็จ! ยอดเงินคงเหลือ: ${self.__balance}")

    def withdraw(self, input_pin, amount):
        if input_pin != self.__pin:
            print("❌ รหัส PIN ไม่ถูกต้อง")
            return

        if amount > self.__balance:
            print("❌ ยอดเงินไม่เพียงพอ")
            return

        print(f"\n--- กำลังทำรายการถอนเงิน: ${amount} ---")
        self.__request_otp()
        if self.__verify_otp():
            self.__balance -= amount
            self.log_history(f"ถอนเงิน: -${amount}")
            print(f"✅ ถอนเงินสำเร็จ ${amount}")

    def transfer(self, target_account, amount, input_pin):
        if input_pin != self.__pin:
            print("❌ รหัส PIN ไม่ถูกต้อง")
            return

        if amount > self.__balance:
            print("❌ ยอดเงินไม่เพียงพอสำหรับการโอน")
            return

        print(f"\n--- กำลังโอนเงิน ${amount} ไปยัง {target_account.account_name} ---")
        self.__request_otp()
        if self.__verify_otp():
            self.__balance -= amount
            target_account.receive_funds(amount, self.account_name)
            self.log_history(f"โอนเงินไปที่ {target_account.account_name}: -${amount}")
            print(f"✅ โอนเงินสำเร็จ!")

    def receive_funds(self, amount, sender_name):
        self.__balance += amount
        self.log_history(f"ได้รับเงินจาก {sender_name}: +${amount}")

    def show_history(self):
        print(f"\n--- ประวัติของคุณ {self.account_name} ---")
        for i, record in enumerate(self.history, 1):
            print(f"{i}. {record}")

# --- การทดสอบระบบ ---
acc_a = ATM("John Wick", "1111", 5000)
acc_b = ATM("Winston", "2222", 1000)

# ลองฝากเงิน (จะมีการถาม OTP)
acc_a.deposit(500)

# ลองโอนเงิน (จะมีการถาม OTP)
acc_a.transfer(acc_b, 1000, "1111")
```

```python
import requests
import time
import pandas as pd

def fetch_anime_batch(id_list):
    results = []

    print(f"เริ่มดึงข้อมูลทั้งหมด {len(id_list)} รายการ...")

    for index, anime_id in enumerate(id_list):
        url = f"https://api.jikan.moe/v4/anime/{anime_id}/full"

        try:
            response = requests.get(url)

            if response.status_code == 200:
                data = response.json()['data']

                # เก็บเฉพาะข้อมูลที่จำเป็นสำหรับการวิเคราะห์ (ปรับแต่งได้ตามใจชอบ)
                anime_info = {
                    'mal_id': data.get('mal_id'),
                    'title': data.get('title'),
                    'type': data.get('type'),
                    'score': data.get('score'),
                    'scored_by': data.get('scored_by'),
                    'rank': data.get('rank'),
                    'popularity': data.get('popularity'),
                    'members': data.get('members'),
                    'favorites': data.get('favorites'),
                    'source': data.get('source'),
                    'genres': ", ".join([g['name'] for g in data.get('genres', [])]),
                    'studios': ", ".join([s['name'] for s in data.get('studios', [])]),
                    'year': data.get('year')
                }
                results.append(anime_info)
                print(f"[{index + 1}/{len(id_list)}] สำเร็จ: {data.get('title')}")

            elif response.status_code == 429:
                print(f"!! เตือน: โดน Rate Limit (429) จะรอ 5 วินาทีแล้วลองใหม่...")
                time.sleep(5)
                # (Option) อาจจะวนซ้ำ ID เดิม หรือข้ามไปก่อน

            else:
                print(f"[{index + 1}/{len(id_list)}] ข้าม ID {anime_id}: Error {response.status_code}")

        except Exception as e:
            print(f"เกิดข้อผิดพลาดที่ ID {anime_id}: {e}")

        # --- ส่วนสำคัญที่สุด: ต้องหน่วงเวลาเพื่อไม่ให้โดนแบน ---
        # Jikan ขอความร่วมมือให้ทิ้งระยะห่างประมาณ 1 วินาทีต่อ request
        time.sleep(1.2)

    return pd.DataFrame(results)

# 1. กำหนดรายการ ID ที่ต้องการ (เช่น จากการไป Scraping มา หรือระบุเอง)
# ตัวอย่าง: ดึงข้อมูล Cowboy Bebop, Naruto, One Piece, Spy x Family
my_ids = [1, 20, 21, 48456]

# 2. รันฟังก์ชัน
df_anime = fetch_anime_batch(my_ids)

# 3. บันทึกเป็นไฟล์ CSV เพื่อนำไปใช้ใน Excel หรือ Power BI
df_anime.to_csv("my_anime_dataset.csv", index=False, encoding='utf-8-sig')
print("\nบันทึกไฟล์ 'my_anime_dataset.csv' เรียบร้อยแล้ว!")
```

```python
import requests
import pandas as pd
import time

def get_top_50_anime():
    top_anime_list = []

    # ดึง 2 หน้าเพื่อให้ได้ 50 อันดับ (หน้าละ 25 รายการ)
    for page in range(1, 3):
        print(f"กำลังดึงข้อมูลหน้า {page}...")
        url = f"https://api.jikan.moe/v4/top/anime?page={page}"

        try:
            response = requests.get(url)
            if response.status_code == 200:
                data = response.json()['data']
                top_anime_list.extend(data)
            else:
                print(f"Error: {response.status_code}")
                break
        except Exception as e:
            print(f"เกิดข้อผิดพลาด: {e}")
            break

        # หน่วงเวลาระหว่างหน้าเพื่อป้องกันการโดนบล็อก
        time.sleep(1.5)

    # แปลงข้อมูลเป็น DataFrame เพื่อการวิเคราะห์
    df = pd.DataFrame(top_anime_list)

    # เลือกเฉพาะคอลัมน์ที่น่าสนใจสำหรับการวิเคราะห์ตลาด
    selected_columns = [
        'rank', 'title', 'score', 'scored_by',
        'type', 'episodes', 'status', 'season', 'year'
    ]

    return df[selected_columns]

# รันฟังก์ชัน
df_top_50 = get_top_50_anime()

# แสดงผล 5 อันดับแรก
print("\n--- 5 อันดับแรกจากข้อมูลที่ดึงมา ---")
print(df_top_50.head())

# บันทึกเป็นไฟล์ CSV สำหรับใช้ใน Power BI
df_top_50.to_csv("top_50_anime.csv", index=False, encoding='utf-8-sig')
print("\nบันทึกไฟล์ 'top_50_anime.csv' สำเร็จ!")
```
