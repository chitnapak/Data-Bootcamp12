````python
import random

def play_game():
    choices = ["ค้อน", "กรรไกร", "กระดาษ"]

    # เพิ่มตัวนับคะแนน
    user_score = 0
    computer_score = 0

    print("--- ยินดีต้อนรับเข้าสู่เกม ค้อน กรรไกร กระดาษ (เวอร์ชันนับคะแนน) ---")
    print("พิมพ์ 'เลิกเล่น' หากต้องการออกจากเกม")

    while True:
        user_choice = input("\nเลือกมาเลย (ค้อน/กรรไกร/กระดาษ): ")

        if user_choice == "เลิกเล่น":
            print("-" * 30)
            print("จบเกม!")
            print(f"คะแนนสุดท้ายของคุณ: {user_score}")
            print(f"คะแนนสุดท้ายของคอมพิวเตอร์: {computer_score}")

            if user_score > computer_score:
                print("สรุปผล: คุณชนะคอมพิวเตอร์ได้ เก่งมาก!")
            elif user_score < computer_score:
                print("สรุปผล: คุณแพ้คอมพิวเตอร์ ไม่เป็นไรนะลองใหม่คราวหน้า")
            else:
                print("สรุปผล: เสมอกันไปอย่างดุเดือด!")
            break

        if user_choice not in choices:
            print("สะกดผิดหรือเปล่า? ลองพิมพ์ใหม่นะ")
            continue

        computer_choice = random.choice(choices)
        print(f"คอมพิวเตอร์เลือก: {computer_choice}")

        # ตรวจสอบผลและบวกคะแนน
        if user_choice == computer_choice:
            print("ผลลัพธ์: เสมอกัน!")
        elif (user_choice == "ค้อน" and computer_choice == "กรรไกร") or \
             (user_choice == "กรรไกร" and computer_choice == "กระดาษ") or \
             (user_choice == "กระดาษ" and computer_choice == "ค้อน"):
            print("ผลลัพธ์: คุณชนะ!")
            user_score += 1
        else:
            print("ผลลัพธ์: คุณแพ้!")
            computer_score += 1

        # แสดงคะแนนปัจจุบัน
        print(f"Score -> คุณ: {user_score} | คอม: {computer_score}")

# เริ่มเกม
play_game()
````

````python
import random

def play_game():
  choices = ["kon", "gungai", "kradad"]

  user_score = 0
  computer_score = 0

  print("welcome to kon, gungai, kradad")
  print("don't to exit")

  while True:
    user_choice = input("kon, gungai, kradad: ")

    if user_choice == "don't":
      print("Thank You for join game")
      print(f"your score: {user_score}")
      print(f"computer score: {computer_score}")
      break

      if user_score > computer_score:
        print("you win")
      elif user_score < computer_score:
        print("you lose")
      else:
        print("draw")
      break

    if user_choice not in choices:
      print("again please (kon, gungai, kradad)")
      continue

    computer_choice = random.choice(choices)
    print(f"computer choice: {computer_choice}")

    if user_choice == computer_choice:
      print("draw")
    elif (user_choice == "kon" and computer_choice == "gungai") or \
         (user_choice == "gungai" and computer_choice == "kradad") or \
         (user_choice == "kradad" and computer_choice == "kon"):
      print("you win")
      user_score += 1
    else:
      print("you lose")
      computer_score += 1

    print(f"your score: {user_score} | computer score: {computer_score}")


play_game()
````
