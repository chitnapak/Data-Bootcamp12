## Group by

```python
# group by +sum/mean
# penguins[penguins['species'] == 'Adelie']['bill_length_mm'].mean()

penguins.groupby('species')['bill_length_mm'].agg(['mean', 'sum', 'std'])
```

```python
# group by more than one columns
result = penguins.groupby(['species', 'island'])['bill_length_mm']\
  .agg(['min', 'mean', 'max'])\
  .reset_index()

result.to_csv('result.csv')
```

## Map

```python
# map values MALE: m ,FEMALE: f
# penguins['sex'].jead()
penguins['sex_new'] = penguins['sex'].map({'MALE': 'm', 'FEMALE': 'f'}).head(10).fillna('other')

penguins.head()
```

## Merge

```python
left = {
    "student_id": ["STU001", "STU002", "STU003"],
    "name": ["James Smith", "Emma Wilson", "Liam Brown"],
    "major": ["Computer Science", "Data Science", "Physics"]
}

right = {
    "student_id": ["STU001", "STU002", "STU004"],
    "math_score": [85, 92, 78],
    "english_score": [88, 95, 80]
}

df_left = pd.DataFrame(left)
df_right = pd.DataFrame(right)
```

```python
df_left
```

```python
import pandas as pd
pd.merge(df_left, df_right, on='student_id')
```
