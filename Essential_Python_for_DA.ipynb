# OOP

* attributes : name,age
* methods: function bark() hbd()

```python
class Dog:
  def __init__(self, name, species, age):
    self.name = name
    self.species = species
    self.age = age

  def bark(self):
    print(f"{self.name} is braking Hong Hong")

  def hbd(self):
    self.age +=1
```

```python
## attributes
dog1 = Dog("milo", "Chihuahua", 3)
dog2 = Dog("davod", "Husky", 2)
```

```python
print(f"{dog1.name} is {dog1.species}")
```

```python
print(f"{dog2.name} is {dog2.species}")
```

```python
dog1.bark()
```

```python
print(dog1.age)
dog1.hbd()
print(dog1.age)
```

```python
## homeworkd OOP for an ATM

## deposit, withdraw, check balance, OTP, order coffee, pay bill, internet etc.
```

```python
class ATM:
  def __init__(self, username, bank, amount):
    self.username = username
    self.bank = bank
    self.amount = amount

  def __str__(self):
    return f"{self.username} uses {self.bank} and has {self.amount} THB in an account."

  def check_balance(self):
    print(f"Balance: {self.amount} THB")

  def deposit(self, saving):
    self.amount += saving
    print(f"Deposit {saving} THB")
    print(f"Balance: {self.amount} THB")
```

```python
toy = ATM("kasidis toy", "scb", 100)
```

```python
toy.check_balance()
```

```python
toy.deposit(1)
```

```python
toy.check_balance()
```

```python
print(toy)
```

```python
class Person:
  def __init__(self, name ,age):
    self.name = name
    self.age = age

## extention
class Employee(Person):
  def __init__(self, name, age, company):
    super().__init__(name, age)
    self.company =company

  def greeting(self):
    print("Hello world")
```

```python
# context manager ไม่ต้องมีเปิดปิด ไฟล์
read_data = []

with open("hotel.csv", "r") as file:
  data = csv.reader(file)
  for row in data:
    read_data.append(row)
```

```python
read_data
```

```python
# header
read_data[0]
```

```python
# body data
read_data[1: ]
```

```python
## for loop to calculate avg. price per night
prices = []

for row in read_data[1: ]:
  prices.append(int(row[4]))

print(f"AVG price per night: {sum(prices)/ len(prices)}")
```

```python
## for loop
import requests
import time

names = []


for i in range(10):
  url = f"https://swapi.info/api/people/{i+1}"
  response = requests.get(url)
  response.json = response.json()
  names.append(response.json["name"])
  time.sleep(1)

print(names)
```

```python
# map / filter
# iterable (adj)
scores = [2, 5, 10, 12]
```
